### Función auxiliar para pruebas de errores

```python
def probar_error(fn, descripcion):
    try:
        resultado = fn()
        print(f"{descripcion} -> {resultado}")
    except Exception as e:
        print(f"{descripcion} -> {type(e).__name__}: {e}")
```
